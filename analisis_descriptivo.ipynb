```python
# vamos comparar variables numericas con la variable objetivo loan_status
numerical_cols = data.select_dtypes(include='number').columns

for col in numerical_cols[4:6]:
    plt.figure(figsize=(10, 6))
    sns.boxplot(x='loan_status', y=col, data=data)
    plt.title(f'Boxplot of {col} by loan_status')
    plt.show()
```

```python
# grafico de densisdad de las variables numericas con respecto a la variable objetivo
for col in numerical_cols[4:6]:
    plt.figure(figsize=(10, 6))
    sns.kdeplot(data=data, x=col, hue='loan_status', fill=True)
    plt.title(f'Density Plot of {col} by loan_status')
    plt.show()
```

```python
# compara variables categoricas con la variable objetivo.
# pintamos un heatmap usand pd.crosstab para cada variable categorica con la variable objetivo loan_status
for col in categorical_cols:
    plt.figure(figsize=(10, 6))
    crosstab = pd.crosstab(data[col], data['loan_status'], normalize=True)
    sns.heatmap(crosstab, annot=True, cmap='Blues', fmt='.2f')
    plt.title(f'Heatmap of {col} by loan_status')
    plt.ylabel(col)
    plt.xlabel('loan_status')
    plt.show()

```
